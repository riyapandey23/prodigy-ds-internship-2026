```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('population.csv')
df.head(5)
```

```python
df.columns
```

```python
df = df[['Country Name','2022']]
df = df.dropna()
top10 = df.sort_values(by='2022', ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=top10['Country Name'],y=top10['2022'])
plt.xlabel('Country')
plt.ylabel('Population')
plt.show()
```

The bar chart shows the top 10 countries by population in 2022. Countries like China and India have the highest population, significantly higher than other countries.

```python
plt.figure(figsize=(10,6))
sns.histplot(df['2022'], bins=10 , kde=True)
plt.title('Population Distribution')
plt.xlabel('Population')
plt.ylabel('Frequency')
plt.show()
```
